The collectors in `cellpy.collect` (used in the `02` - `04` notebooks) return a collection.
`collection.save("out")` writes the collected **frame** (`parquet` + `csv`, plus a
`meta.json`), while `collection.plot(...)` returns the figure. Use
`plotutils.save_image_files(fig, name)` to write the **figure** itself as `png` / `svg` /
plotly `json` (needs `kaleido`), or `collection.to_image("png")` if you only want the bytes.
A figure written as plotly `json` can be read back here and tweaked without re-collecting.
